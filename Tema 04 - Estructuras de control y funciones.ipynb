###  1. El condicional `if`

Como ya hemos visto, la instrucción `if` nos permite ejecutar distIntos bloques de instrucciones dependiendo de que se cumpla una determinada condición o no.

La estructura general de una instrucción if es la siguiente:

```python
if <CONDICIÓN 1>:
    <BLOQUE 1>
elif <CONDICIÓN 2>:
    <BLOQUE 2>
elif <CONDICIÓN 3>:
    <BLOQUE 3>
....
else:
    <BLOQUE n>
```

Esta instrucción se interpreta como sigue:
* Si `<CONDICIÓN 1>` se cumple, se ejecuta la secuencia de instrucciones de `<BLOQUE 1>`.
* Si no se cumple `<CONDICIÓN 1>` y se cumple `<CONDICIÓN 2>`, se ejecuta la secuencia de instrucciones de `<BLOQUE 2>`.
* Si no se cumple `<CONDICIÓN 1>` ni `<CONDICIÓN 2>`, pero sí `<CONDICIÓN 3>`, se ejecuta la secuencia de instrucciones de `<BLOQUE 3>`.
* En general, si se cumple la condición de una línea `elif` y no se cumplen las condiciones de las anteriores, se ejecuta el bloque de instrucciones que aparece a continuación de ese `elif`.  
* Si no se cumple ninguna de las condiciones, entonces se ejecuta `<BLOQUE n>`

Tanto los `elif` como el `else`son **opcionales**. En caso de aparecer un `else`, este debe ser el último bloque. 

**Ejemplo:** En el siguiente ejemplo, podemos ver una función que recibe una calificación numérica y devuelve el correspondiente literal:

````python
def calificacion_literal(calificacion):
    if calificacion >= 9:
        return "Sobresaliente"
    elif calificacion >= 7:
        return "Notable"
    elif calificacion >= 5:
        return "Aprobado"
    else:
        return "Suspenso" #cuando hay un return hasta aquí hemos llegado, se para la función
````

**Ejemplo:** la siguiente función nos sirve para calcular la puntuación de un examen tipo test  en el que cada respuesta acertada vale 1 punto, y cada error resta medio punto. Cada respuesta puede ser `"a"`, `"b"`, `"c"`, `"d"` ( o `"-"` si no se contesta).

````python
def puntuacion(respuestas,soluciones):
    acum=0
    for i in range(len(respuestas)): #para cada i de la lista de la longitud es decir de 0 al final -1/ [0,...,Len()-1]
        r=respuestas[i]
        s=soluciones[i]
        if r!="-":
            acum+=(1 if r==s else -0.5)
    return max(0,acum)#acota por abajo a 0 porque no puedes sacar un -1
````

````python
puntuacion(["a","b","a","c","d","-","d","b","c","b"],
           ["a","b","c","c","d","a","b","d","c","b"])
````

````python
puntuacion(["a","b","a","c","-","-","d","-","c","b"],
           ["a","b","a","d","d","c","d","b","c","b"])
````

Nótese la diferencia entre la instrucción `if` que en el ejemplo sirve para evitar las respuestas no contestadas, y la expresión `if`, que en el ejemplo se evalúa a un número (`1` ó `-0.5`), dependiendo de una condición. 

Podríamos haber definido incluso la función anterior sólo con expresiones `if`, aunque hubiera sido en detrimento de la claridad del código:

````python
def puntuacion_2(respuestas,soluciones):
    acum=0
    for i in range(len(respuestas)):
        ri=respuestas[i]
        si=soluciones[i]
        acum+=(0 if ri=="-" else (1 if ri==si else -0.5))
    return max(0,acum)
````

## 2. Bucle `while`

Los bucles `while` nos van a permitir ejecutar repetidamente (en *bucle*) un bloque de instrucciones, mientras se siga cumpliendo una determinada condición. Entendemos por *iteración* cada una de las repeticiones que se realizan en un bucle. 

Un bucle `while` tiene la siguiente forma:

```python
while <CONDICIÓN>: #el while tmb es comprobacion devuelve un true o un false
    <BLOQUE 1>
else:
    <BLOQUE 2>
```

La cláusula `else` es opcional, y en la práctica se usa poco.


La ejecución de un bucle `while` se produce de la siguiente manera:

1. Se evalúa `<CONDICIÓN>`. 
 * Si es verdadera, se ejecuta `<BLOQUE 1>` y se vuelve al punto 1
 * Si no es verdadera, se da por terminado el bucle. 
2. Si el bucle se termina normalmente, y hay cláusula `else`, ejecutar `<BLOQUE 2>`.

Más adelante especificaremos qué se entiende por "terminar normalmente" el bucle, aunque reiteramos que en la práctica el bloque `else` de un `while` no se suele usar. 

Veamos algunos ejemplos de instrucciones `while`. 

**Ejemplo:** La siguiente función recibe una lista de números `l` y una cota `C`, y devuelve **la posición** del primer elemento de `l` cuya suma acumulada (es decir, la suma de ese elemento con los anteriores en la lista) supere a `C` (o la última posición si no se supera para ningún elemento):

````python
def pos_acum_cota(l,C):
    i=-1
    acum=0
    while acum <= C and i<len(l)-1: #mientras que no llegue a la cota sigue para adelante, combinancion de operaciones logicas comporbacion
        i+=1
        acum+=l[i] # acumula los valores de la lista mientras la suma no supere a C
    return i       # si la supera, sale del bucle while y devuelve el valor de i en ese momento
````

````python
pos_acum_cota([3,-2,6,4,-8,13,2,-5,11,7,8],15)
````

````python
pos_acum_cota([3,-2,6,4,-8,13,2,-5,11,7,8],55)
````

**Ejemplo:** La siguiente función pide números sucesivamente por pantalla, y finalmente devuelve la media aritmética de los números introducidos.  

````python
def calcula_media():
    acum=0
    cont=0
    seguir="s"
    while seguir=="s":
        x=float(input("Introducir un número: "))
        acum+=x
        cont+=1
        seguir=input("¿Más números? (sí [s] o no [no])")
    print(f"\nLa media de los números introducidos es: {acum/cont}.")
````

**Ejemplo:** Un último ejemplo que usa `while`. La siguiente función calcula la raiz cuadrada de un número natural `n` mayor que 1, con una cota de precisión dada `epsilon`, mediante el método de búsqueda dicotómica.

````python
def raiz_cuadrada(n,epsilon):  # suponemos n>=2
    inf=1                      # Buscamos la raiz en el intervalo (inf,sup)     
    sup=n                      # Inicialmente el intervalo es (1,n)
    while (sup-inf)>epsilon:   # Mientras el intervalo sea mayor que la precisión
        med=(inf+sup)/2        # Calculamos el punto medio med
        if med**2>n:           # Si su cuadrado es mayor que n, la raiz está en (inf,med)  
            sup=med            
        elif med**2<n:         # Si su cuadrado es menor que n, la raiz está en (med,sup)
            inf=med
        else:                  # Si el cuadrado es n, la raiz está en (med,med)
            inf=med
            sup=med
    return med
        
````

````python
raiz_cuadrada(2,0.000001)
````

````python
raiz_cuadrada(3,0.000000000001)
````

Mostremos algunos ejemplos del uso de `break`.

**Ejemplo:** El código de la función anterior `raiz_cuadrada` queda más claro y eficiente si se ejecuta un `break` en el caso de que se encuentre la raiz de manera exacta:

````python
def raiz_cuadrada_2(n,epsilon):  # suponemos n>=2
    inf=1                      # Buscamos la raiz en el intervalo (inf,sup)     
    sup=n                      # Inicialmente el intervalo es (1,n)
    while (sup-inf)>epsilon:   # Mientras el intervalo sea mayor que la precisión
        med=(inf+sup)/2        # Calculamos el punto medio med
        if med**2>n:           # Si su cuadrado es mayor que n, la raiz está en (inf,med)  
            sup=med            
        elif med**2<n:         # Si su cuadrado es menor que n, la raiz está en (med,sup)
            inf=med
        else:                  # Si el cuadrado es n, la raiz es med y salimos del bucle
            break
    return med
        
````

````python
raiz_cuadrada_2(10,0.000001)
````

**Ejemplo:** Otro ejemplo interesante del uso de `break` lo obtenemos reformulando la función `calcula_media_2`que vimos anteriormente:

````python
def calcula_media_3():
    acum=0
    cont=0
    while True:
        x_str=input("Introducir un número (Intro para terminar): ")
        if x_str!="":
            acum+=float(x_str)
            cont+=1
        else:
            break
    print(f"\nLa media de los números introducidos es: {acum/cont}")
````

````python
calcula_media_3()
````

Nótese el uso de `True` como condición del bucle `while`. Aunque en principio pueda parecer que estamos provocando un bucle infinito, la salida del bucle la controla la condición del `if` y el `break` en el caso de que no se cumpla.

* La instrucción `continue` ejecutada dentro de bucle, termina la iteración actual, pasando a la siguiente iteración sin ejecutar el resto de instrucciones del bloque.

**Ejemplo:** La siguiente función imprimiría sólo los números impares menores o iguales que un número dado `n`:

````python
def impares(n):
    num=0
    while num < n:
        num+=1
        if (num % 2) == 0:
            continue
        print(num)
````

````python
impares(15)
````

* Por último, como ya hemos visto, la instrucción `return` sirve para devolver el valor o valores calculados por una llamada a una función. Por tanto la ejecución de un `return` dentro de un bucle, termina la ejecución de la función, interrumpiendo no solo el bucle sino todas las instrucciones a continuación del bucle. Veámoslo con un ejemplo:

**Ejemplo:** La siguiente función `f`recibe una lista de números naturales y va imprimiendo los números impares de la lista hasta que se encuentra un número par: 

````python
def f(l):                  
    i=0
    while i<len(l):
        if l[i]%2==0:      # Detectado un número par en l
            print(f"Número par {l[i]} en la posición {i}. STOP")
            return
        print(f"Número impar {l[i]} en la posición {i}.")
        i+=1
    print("La lista no tiene números pares")
````

````python
f([1,3,7,10,11,13])
````

Si una función `g` llama a otra función `f`, un `return` dentro de `f` interrumpe la ejecución de `f` y devuelve en control a la siguiente instrucción de `g` tras la llamada a `f`. Por ejemplo: 

````python
def g():
    print("Primera línea de g")
    f([1,3,7,10,11,13])
    print("Tercera línea de g")
````

### 2.2. La cláusula `else` en un bucle while

Como hemos visto al principio, el bucle `while` en Python incluye la posibilidad de una cláusula `else` opcional (no confundir con el `else` de una instrucción `if`).

El bloque de instrucciones de una cláusula `else` se ejecuta si el bucle ha terminado *normalmente*. Es decir, porque la condición del `while` ha dejado de cumplirse; o dicho de otra manera, no termina debido a un `break`. 

**Ejemplo:** La siguiente función recibe un elemento `x` y una lista `l`, y devuelve la posición de la primera ocurrencia de `x` en `l`, o -1 en caso de que `x` no esté en `l`.

````python
def pos_x(x,l):
    i=0
    while i<len(l):
        if l[i]==x:
            break
        i+=1
    else:
        i=-1    # Sólo se ejecutará esta instrucción si el bucle no se termina con un break
                # Eso sólo ocurre si x no está en l
    return i
````

````python
pos_x("a",["c","d","a","c","a","f"])
````

````python
pos_x("x",["c","d","a","c","a","f"])
````

## 3. Bucle `for`

El bucle `for` puede verse como un caso particular de una estructura de la estructura `while`, en la que las iteraciones se realizan tomando los sucesivos elementos de una secuencia. A pesar de eso, la versatilidad de los bucles `for`en Python hace que en la práctica sean más usados que los bucles `while` para implementar una estructura de control repetitiva. 

A pesar de que ya hemos visto en los temas anteriores un uso básico de bucles `for`, vamos a verlos en esta sección con más profundidad.  

La sintaxis de un bucle `for` es la siguiente:

```python
for <EXPR> in <ITERABLE>:
    <BLOQUE 1>
else:
    <BLOQUE 2>
```

Al ejecutarse el `for`, se ejecutará repetitivamente la secuencia de instrucciones de `<BLOQUE 1>`, donde `<EXPR>` irá tomando los valores de `<ITERABLE>` en las sucesivas iteraciones. Al igual que en el bucle `while`, la clásula `else` de un bucle `for` es opcional. Si hay cláusula `else` y el bucle termina normalmente (es decir, se recorre completamente el `<ITERABLE>` sin que se ejecute un `break`), entonces antes de finalizar ejecuta la secuencia de instrucciones del `<BLOQUE 2>`. 



Sin entrar en detalles técnicos, un *iterable* en Python es cualquier dato que puede ser recorrido en secuencia. Ya hemos visto varios tipos de datos que son iterables en Python: listas, strings, tuplas y rangos. Hay más tipos de datos iterables, como los diccionarios, los conjuntos, los ficheros y los generadores, algunos de los cuales veremos en los temas siguientes.

Usualmente, `<EXPR>` es una variable individual, pero por ejemplo podría ser también una tupla de variables, si los elementos del iterable fueran tuplas. 

Todo lo que se ha descrito anteriormente sobre `break`, `continue` y `return` para bucles `while` se aplica exactamente igual en bucles `for`.

Veamos a continuación algunos ejemplos de bucles `for`.

**Ejemplo:** La siguiente función recibe una lista de números y devuelve la posición en la lista del mayor de esos números (la primera si el máximo ocurre más de una vez):

````python
def pos_máximo(l):
    mayor=-float("inf") #hace referencia infinito
    pos=0
    for x in l:
        if x>mayor:
            mayor=x
            pos_mayor=pos
        pos+=1
    return pos_mayor, mayor
````

**Ejemplo:** La siguiente función recibe una secuencia no vacía `seq_str` (lista o tupla) cuyos elementos son strings, y un string `sep`, y devuelve un string resultado de concatenar los elementos de `seq_str` intercalando `sep` entre ellos (es decir, similar al método `join` de los strings):

````python
def mi_join(seq_str,sep):
    res=seq_str[0]
    for s in seq_str[1:]:
        res+=(sep+s)
    return res
````

**Ejemplo:** La siguiente función recibe un número natural `n` (que supondremos mayor que 1)  y devuelve un lista con los `n+1` primeros números de la sucesión de Fibonacci:

````python
def seq_fibonacci(n):
    res=[0,1]
    for i in range(2,n+1):
        res.append(res[i-2]+res[i-1])#cada vez que avance un numero va mirando los dos anteriores y los va sumando
    return res
````

### 3.1. Bucles anidados

Un *bucle anidado* es un bucle que forma parte del bloque de instrucciones de otro bucle. Al bucle que está dentro de otro le llamamos *bucle interno*, y al que lo contiene le llamamos *bucle externo*. Puede incluso haber varios niveles de anidamiento. 

En programación es muy común el uso de bucles anidados, ya que es la opción natural es muchas situaciones. Por ejemplo, para recorrido de estructuras bidimensionales (como las matrices), o para iteraciones que a su vez requieran en cada iteración una estructura repetitiva. Vemos algunos ejemplos:

**Ejemplo:** La siguiente función, dados dos números naturales positivos `i`y `j`, imprime los números entre 1 y `i*j` en una tabla de `i` filas y `j` columnas:

````python
def imprime_tabla_num(i,j):
    for f in range(1,i+1):
        for c in range(1,j+1):
            num=j*(f-1)+c
            print(f"{num:>5}",end="") # Formateo: 5 espacios por columna, justificado a la derecha,  sin salto de línea
        print()   # Salto de línea al final de cada fila                        
````

````python
imprime_tabla_num(10,17)
````

En la función anterior, la variable `f` indica el número de fila, y la `c` el número de columna. En el bucle externo, la fila `f` va tomando valores desde 1 hasta 'i', de ahí el rango hasta `i+1`. Para cada valor de `f`, se ejecuta el bucle interno, variando la columna `c` desde 1 hasta 'j', de ahí el rango `j+1`.  

**Ejemplo:** La siguiente función, dado un número natural `n` mayor que 1, devuelve una lista con los números primos menores que `n`. Además imprime por pantalla una factorización no trivial de los números no primos en ese mismo rango.

````python
def mat_escalar(m,k):
    res=[]
    for f in m:
        fila_res=[]
        for x in f:
            fila_res.append(x*k)
        res.append(fila_res)
    return res
````

## 4. Excepciones y manejo de errores

- **`ValueError`**: aplicación de operaciones o funciones sobre algo de tipo adecuado pero valor inadecuado.
- ...

Por defecto, cuando durante la ejecución se comete un error, el programa se detiene y devuelve el correspondiente mensaje informando del error.

### 4.1. La construcción `try .. except ...`

Con `try ... except ...` tenemos la posibilidad de *gestionar* un error. Esto significa que cuando un error de un determinado tipo ocurre durante la ejecución, en lugar de detenerla, se desvía el control hacia una serie de instrucciones que efectúan una serie de acciones encaminadas a recuperarse del error, sin necesidad de parar. 

La sintaxis es la siguiente:
```python
try:
    <BLOQUE_TRY>
except [<EXCEPCIÓN(ES)>]:
    <BLOQUE_EXCEPT>


Esto es lo que ocurre cuando se ejecuta un `try ... except ...`:

* Primero, se ejecuta `<BLOQUE_TRY>`

* Si no ocurre ninguna excepción, el bloque tras `except` se omite y la ejecución del `try` finaliza.

* Si ocurre una excepción durante la ejecución de  `<BLOQUE_TRY>`, se omite el resto de ese bloque. Si el tipo de excepción coincide con la excepción (o excepciones) nombradas en <EXCEPCIÓN(ES)>, entonces se ejecuta  `<BLOQUE_EXCEPT>` (es lo que se llama *atrapar* o *gestionar* la excepción), y luego la ejecución continúa después del bloque `try ... except ...`. **Nota:** En el caso particular de que después del `except` no aparezca ningún nombre de excepción, se atrapa cualquier excepción, independientemente de su tipo.  

* Si ocurre una excepción que no coincide con lo indicado en el `except` (y no se está dentro de ningún otro bloque `try` que gestione la excepción), la ejecución se interrumpe con el correspondiente mensaje informativo.


**Ejemplo:** La siguiente función pide un número al usuario y devuelve el doble de ese número. Tal y como está definida, si el usuario introduce algo que `int`no puede transformar en un número, va a devolver un error de tipo `ValueError` (recordemos que `input` devuelve lo que teclea el usuario como un string).

````python
def devuelve_doble():
    x= int(input("Introduzca un número: "))
    return 2*x

#devuelve_doble()

#Introduzca un número: a
#---------------------------------------------------------------------------
#ValueError                                Traceback (most recent call last)
# <ipython-input-3-4e27c5441a1d> in <module>
#       3     return 2*x
#       4 
# ----> 5 devuelve_doble()#
#
# <ipython-input-3-4e27c5441a1d> in devuelve_doble()
#       1 def devuelve_doble():
# ----> 2     x= int(input("Introduzca un número: "))
#       3     return 2*x
#       4 
#       5 devuelve_doble()
#
# ValueError: invalid literal for int() with base 10: 'a'
````

Podemos modificar la función para gestionar ese posible error, mediante `try ... except ...`:

````python
def devuelve_doble_2():
    while True:
        try:
            x= int(input("Introduzca un número: "))
            return 2*x
        except ValueError:
            print("No es un número, inténtelo de nuevo.")

devuelve_doble_2()
````

Nótese que la función usa un while con una condición `True`, y que por tanto se sale del bucle mediante la instrucción `return 2*x`. En el caso de que se produzca una excepción de tipo `ValueError`, el bloque `except` imprime un mensaje y el cuerpo del bucle `while` vuelve a ejecutarse. El efecto que esto tiene en la ejecución es que se nos pedirá que introduzcamos un número hasta que efectivamente se teclee un número. 

### 4.2. Usando `try ... except ...` para controlar la ejecución

No necesariamente el uso de `try ... except ...` es adecuado para gestionar errores, sino que a veces es un simple mecanismo de control similar a `if`, pero que en ocasiones puede resultar más eficiente. Veámoslo con un ejemplo:

**Ejemplo:** Supongamos que tenemos una lista de tuplas cuyos elementos son de la forma $(n,x)$ donde $n$ es un número natural positivo (la llamaremos la versión "comprimida" de la lista). La siguiente función transforma ese tipo de listas en otra lista en la que cada tupla se sustituye por el elemento $x$ repetido $n$ veces. 

````python
def expande(l):
    res=[]
    for n,x in l:
        res.extend([x]*n)
    return res
````

````python
expande([(3,"a"),(4,"b"),(2,"c"),(5,"d")])
````

Supongamos ahora que para tuplas de la forma $(1,x)$, queremos poner en la lista comprimida simplemente $x$. En ese caso, la función anterior nos daría un error de ejecución, ya que la definición anterior está suponiendo que todos los elementos de la lista son pares:

````python
# expande([(2,"a"),(4,"b"),"c",(7,"d")])

#---------------------------------------------------------------------------
#ValueError                                Traceback (most recent call last)
#/tmp/ipykernel_5187/827362004.py in <module>
#----> 1 expande([(2,"a"),(4,"b"),"c",(7,"d")])

#/tmp/ipykernel_5187/2286470289.py in expande(l)
#      1 def expande(l):
#      2     res=[]
#----> 3     for n,x in l:
#      4         res.extend([x]*n)
#      5     return res

#ValueError: not enough values to unpack (expected 2, got 1)
````

Para arreglarlo, podemos rehacer la definición, e introducir un condicional que compruebe para cada elemento si es una tupla o no:

````python
def expande_2(l):
    res=[]
    for p in l:
        if type(p) is tuple and len(p)==2: # type devuelve el tipo de un dato
            res.extend([p[1]]*p[0])
        else:
            res.append(p)
    return res
````

````python
expande_2([(2,"a"),(4,"b"),"c",(7,"d")])
````

Alternativamente, podemos usar `try .. except ..` para tratar de manera diferenciada las tuplas de los elementos individuales:

````python
def expande_3(l):
    res=[]
    for p in l:
        try:
            res.extend([p[1]]*p[0])
        except IndexError:
            res.append(p)
    return res
````

````python
expande_3([(2,"a"),(4,"b"),"c",(7,"d")])
````

En la definición de `expande_3`, estamos suponiendo que los elementos de `l` son tuplas. Cuando uno de los elementos no es una tupla, entonces al acceder a sus componentes `p[0]` y `p[1]` se lanzará una excepción `IndexError`, que será atrapada por el `except` y se ejecuta la instrucción correspondiente a los elementos que nos son tuplas. 

En general, si se espera que usalmente los elementos de la lista sean pares, y de manera infrecuente elementos individuales, entonces la versión `expande_3` con manejo de excepciones es más eficiente que la versión `expande_2` con el condicional `if`, que comprobaría el tipo en cada elemento.  

### 4.3. Lanzando excepciones propias

En un momento dado, como por ejemplo, dentro de una función, puede ser interesante no solo capturar posibles excepciones que ocurran en nuestro código, sino lanzar una excepción. Esto es útil en los casos en los que queramos que nuestra función lance algún error, sin que ello signifique que muestre algo por pantalla y pare la ejecución, sino que simplemente indique dicho error, que luego el programa principal que llame a la función tenga en cuenta como desee.

Para ello, usaremos la palabra `raise` seguida del tipo de excepción a enviar. En el caso más general, podemos lanzar una excepción genérica, de tipo `Exception`. Además, como se ve en el siguiente ejemplo, entre paréntesis puede recibir un *string* con un mensaje de error propio.

Hay que tener en cuenta además, que el hecho de lanzar un error con `raise`, hace que la ejecución de la función en cuestión termine.

### 5.2. Empaquetado y desempaquetado de argumentos  

En Python, el símbolo `*` además de para multiplicar o para copia repetida, se puede usar para *empaquetar* o *desempaquetar* (*unpacking*) argumentos, tanto en la llamada como en la definición de funciones.

El desempaquetado tiene lugar cuando una función, que espera una serie de argumentos individuales, puede recibirlos de manera conjunta en forma de tupla. 

**Ejemplo:** Volvamos a mostrar la definición de la función `j`anterior:

````python
def j(x,y,z=0): 
    return(x**y + z)
````

Supongamos que queremos llamar a la función `j` sobre los argumentos `x=2`, `y=3`, `z=4`. Entonces la llamada debería ser `j(2,3,4)`. Pero imáginemos que todos o parte de los argumentos los tenemos en una tupla o en una lista. En ese caso, es posible usar el operador prefijo `*` para desempaquetar los argumentos en **la llamada** a la función. Veámoslo con ejemplos:

````python
l=(2,3,4)
# j(l) <--- Esta llamada es errónea
j(*l)  # x=2, y=3, z=4
````

````python
m=[4,8]
j(2,*m) # x=2, y=4, z=8
````

````python
j(*m)   # x=4, y=8, z=0 (valor por defecto)
````

El operador de desempaquetado se puede usar **al definir** una función. En ese caso, el efecto que tiene es poder llamar a la función con los argumentos individuales, aunque en el cuerpo de la función serán considerados como si fueran elementos de una tupla: 

**Ejemplo:** Lo siguiente define una función con un número variable de argumentos, que devuelve el producto de los mismos.

````python
def producto(*args):   # args es la tupla que recoge los argumentos en la llamada a la función
    res = 1
    for arg in args:
        res *= arg
    return res
````

````python
producto(2,3)
````

````python
producto(4,2,4,5,6,72,3,9)
````

````python
producto(2)
````

````python
producto()
````

**Ejemplo:** La siguiente función combina desempaquetado con un argumento por defecto:

````python
def suma_de_potencias(*args, pot=1):
    res = 0
    for arg in args:
        res += arg ** pot
    return res
````

````python
suma_de_potencias(2,3,4,2,1,3) # 2+3+4+2+1+3
````

### 5.3. Variables globales y locales

En Python, las variables que se definen y usan dentro de una función se denominan *variables locales*. Estas variables están disponibles para la ejecución de la función, pero no fuera de ese alcance local. Por ejemplo: 

````python
def f(n):
    h=0
    res=0
    for y in range(n):
        res+=y
        h+=1
    return res
# h,y,res son variables locales a la función f    
````

````python
f(8)
# h   # Descomentar la línea para ver el error: "name 'h' is not defined"
````

Es posible definir *variables globales* que pueden ser accedidas por varias funciones de un mismo programa. 

**Ejemplo:** Las siguientes funciones `f1`y `f2` acceden ambas a la variable global `mi_var`. Sin embargo, la variable `aux`es local en ambas definciones:

````python
mi_var=3

def f1(x):
    aux=x*x
    return aux+mi_var

def f2(x):
    aux=3*x
    return aux*mi_var
    
````

````python
f1(8) # 8*8+3
````

````python
f2(5) # 3*5*3
````

En general, **el uso de variables globales no es buena práctica de programación**, ya que el hecho de que el resultado de una función dependa de un valor externo a ella, puede ocasionar comportamientos inesperados  de difícil depuración. Por ejemplo:

**Ejemplo:** Al depender de una variable externa al cuerpo de la función, una misma llamada a la función puede dar dos resultados distintos:

````python
f1(6)
````

````python
mi_var=7
````

````python
f1(6)
````

Sin embargo, hay veces que el uso de variables globales puede ser razonable, siempre que contengan valores que son considerados como *constantes globales* que pueden ser accedidas por distintas funciones. Por ejemplo:  

**Ejemplo:** Podemos guardar el valor de la constante `pi`, que puede ser usada tanto para calcular la longitud de  una circunferencia, como el aŕea de un círculo.

````python
pi=3.14159

def long_circunferencia(radio):
    return 2*pi*radio

def area_circulo(radio):
    return pi*radio**2
````

````python
long_circunferencia(3)
````

````python
area_circulo(4)
````

### 5.4. Funciones anidadas

Por *función anidada* entendemos aquella que cuya definición está dentro de la definción de otra función (que llamaremos *externa*). El uso de funciones anidadas puede servir en muchos casos para tener un código más modular y compacto, evitando repeticiones de líneas de código. 

**Ejemplo:** la siguiente función `descuenta` calcula la suma total de ingresos por las compras de una serie de clientes, algunos de los cuales tienen un descuento por socio. Cada compra viene representada por un par con el nombre del comprador y el precio del artículo. La función `aplica_descuento` reduce el precio de la compra con el corresondiente descuento, a aquellos socios compradores. 

````python
def descuenta(lista_socios,compras):
    
    def aplica_descuento(precio,descuento=1.0):
        return precio*descuento
    
    total= 0
    for nombre,pr in compras:
        if nombre in lista_socios:
            total+=aplica_descuento(pr,descuento=0.7)
        else:
            total+=aplica_descuento(pr)
    return total        
        
````

````python
lista_socios=["Juan","Luis","Arturo"]

descuenta(lista_socios,[("Pedro",214),("Luis",125),("Juan",60),("Miguel",30),("Luis",67),("Arturo",35)])
````
