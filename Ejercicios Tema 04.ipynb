### Ejercicio 1

La siguiente figura representa un *árbol de decisión* que indica si un día es adecuado o no para jugar al tenis, en función de una serie de características.



Suponiendo que las características de un día concreto las representamos mediante una tupla de longitud 4, en la que:

* La primera posición indica el estado del **Cielo**, con posibles valores *Soleado*, *Nublado* y *Lluvia*.
* La segunda componente indica la **Humedad**, con posibles valores *Alta* o *Normal*.
* La tercera posición indica el **Viento**, con posibles valores *Fuerte* o *Débil*
 
Definir una función `arbol_decision_jugar_tenis(tupla)` que recibiendo una tupla con las características de un día concreto, devuelve `sí` o `no` acorde con lo indicado en el árbol de la imagen   

```python
# Solución:
ruptor = 0
for i in range (3):
    if(i=0):
        print('Que tipo de cielo tenemos hoy?')
        print('1. Lluvia\n2. Soleado \n3. Nublado')
        aux1=input('Introduce 1, 2 o 3 ')
        while(ruptor == 0):
            if(aux1==1):
                cielo='Luvia'
                ruptor = 1
            elif(aux1==2):
                cielo='Soleado'
                ruptor = 1
            elif(aux1==3):
                cielo='Nublado'
                ruptor = 1
            else():
                aux1= input("Número introduci invalido. Introduce otro: ")
    else():
        print('hola')








```

```python
# Ejemplos:
dia1=("Lluvia","Alta","Débil")
dia2=("Soleado","Alta","Fuerte")
dia3=("Nublado","Alta","Débil")

# >>> print(arbol_decision_jugar_tenis(dia1))
# si

# >>> print(arbol_decision_jugar_tenis(dia2))
# no

# >>> print(arbol_decision_jugar_tenis(dia3))
# si
```
