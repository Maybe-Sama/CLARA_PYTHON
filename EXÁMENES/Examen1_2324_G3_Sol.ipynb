El determinante $|A|$ de una matriz cuadrada $A$ de tamaño 1 se define como el valor de dicha celda de la matriz.
$$ Si \ A = (a); \ |A|=a $$

El determinante $|A|$ de una matriz cuadrada $A$ de tamaño 2 se define como el producto de los elementos de la diagonal principal, menos el producto de los elementos de la diagonal secundaria:

$$
Si \ A = \left(
\begin{array}{ll}
a_{00} & a_{01} \\
a_{10} & a_{11}
\end{array}
\right) ; \ 
|A| = (a_{00} \cdot a_{11}) - (a_{10} \cdot a_{01})
$$

Para matrices de mayor tamaño, el determinante se puede calcular por la *regla de Laplace*, haciendo por ejemplo el desarrollo por una de las filas. De forma general, el determinante $|A|$ de una matriz cuadrada $A$ de tamaño *n* se calcula como:

$$ |A|=\sum_{j=0}^{n-1} a_{ij} \cdot (-1)^{i+j} \cdot |M_{ij}|$$

, suponiendo un valor fijo de $i$, que será la fila escogida; $a_{ij}$ será cada elemento de la fila escogida, y $|M_{ij}|$ es el menor *ij*, es decir, el determinante de la matriz resultante de eliminar la fila *i* y la columna *j* de $A$. El operando $(-1)^{i+j}$ permite alternar el signo de la suma de elementos.

Como ejemplo concreto, y de forma ilustrativa, consideremos la siguiente matriz de tamaño 3:

$$
A = \begin{pmatrix}

\end{pmatrix}
$$

Supongamos un diccionario que contiene un listado de películas. Las claves del diccionario corresponden al título de la película, y el valor asociado a cada una de ellas es una tupla con 3 valores: el año de estreno, el género, y una lista de valoraciones.

```python
peliculas = {
    'Eduardo Manostijeras': (1990, 'Fantasía', [7.9, 7.7, 8.9]),
    'Harry Potter': (2001, 'Fantasía', [8.8, 6.8, 8]),
    'El Señor de los Anillos': (2001, 'Fantasía', [8.8, 8, 9.6]),
    'La Novia Cadáver': (2005, 'Musical', [7.2, 8.6, 8]),
    'Elemental': (2023, 'Animación', [6.5, 7.6]),
    'Star Wars Episodio IX': (2019, 'Acción', [5.7, 7]),
    'Los Vengadores': (2012, 'Acción', [8, 6.9, 8]),
    'Emoji': (2017, 'Animación', [3.5, 4, 4.8]),
    'La Mansión Encantada': (2003, 'Fantasía', [4.2, 6.4, 5.3]),
}
```

Implemente una función que, para cada género distinto de las películas favoritas del usuario (sin tener en cuenta repetición de géneros), recomiende la mejor película de cada género (que no esté ya entre las favoritas del usuario), siempre y cuando supere un umbral dado. (Ver los ejemplos posteriores).
 * Por ejemplo, si la película con mejor valoración de un género ya está entre las favoritas del usuario, debe recomendarle la siguiente con mejor valoración (y así sucesivamente); si no quedan películas de ese género, no incluye ninguna recomendación para ese género.
 * Si entre las películas favoritas del usuario hay varias de un mismo género, el sistema solo debe recomendar una película para ese género.

La función recibirá los siguientes parámetros:
 * `peliculas`: Un diccionario con la información de las películas, como el de la anterior celda. La función no debe modificarlo.
 * `favoritas`: Una lista de *strings* con los títulos de películas que le han gustado al usuario (se presupone que dichas películas están en el diccionario). La función no debe modificarla.
 * `recomendaciones`: Una lista donde se almacenarán las recomendaciones; será una lista de títulos de películas. La función debe modificar dicha lista para incluir las nuevas recomendaciones; por tanto, no es necesario que la función devuelva nada, solo modificar esta lista. 
 * `umbral`: Valor numérico. No se tendrán en cuenta para la recomendación películas cuya valoración media esté por debajo de dicho umbral. Será un parámetro opcional; si no se indica nada, el umbral será de 5. Supongamos por ejemplo que la mejor película para un género dado está por debajo de dicho umbral; en ese caso, no se incluirá ninguna película recomendada para ese género.

```python
# Solucion: 
def recomendar_pelicula(peliculas, favoritas, recomendaciones, umbral=5.0):
    generos_a_recomendar = set()
    for pelicula in favoritas:
        generos_a_recomendar.add(peliculas[pelicula][1])
    
    for genero in generos_a_recomendar:
        max_valoracion = 0
        max_pelicula = None
        for pelicula in peliculas:
            med = sum(peliculas[pelicula][2])/len(peliculas[pelicula][2])
            if peliculas[pelicula][1] == genero and pelicula not in favoritas and med >= umbral and med > max_valoracion:
                max_valoracion = med
                max_pelicula = pelicula

        if max_pelicula != None:
            recomendaciones.append(max_pelicula)
    
```
