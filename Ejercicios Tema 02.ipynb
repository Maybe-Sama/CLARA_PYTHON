La entropía de Shannon es una función que aparece en teoría de la información. Se define como
$ E = - \sum_{i=1}^n p_i \log p_i $, donde $p$ es un vector de probabilidad, es decir, un vector con coordenadas positivas tal que $\sum_{i=1}^n p_i = 1$. Define una función `shannon(xs)` que compruebe si la lista `xs` representa un vector de probabilidad y, en caso favorable, devuelva la entropía de Shannon. En caso contrario debe devolver `math.nan` (constante que indica que un valor no es un número: *NaN: not a number*).

```python
# Solución:
```

```python
# Ejemplos:

# >>> shannon([0.2, 0.1, 0.2, 0.2])
# nan

# >>> shannon([1/4, 1/4, 1/2])
# 1.0397207708399179

# >>> shannon([1/3, 1/3, -1/3])
# nan
```
